```python
produtos = {
    "mouse":98.55,
    "teclado":30.00,
    "monitor":180.90,
    "gabinete":148.90,
    "hd externo": 890.00,
    "headset": 121.10
}

while True:
    produto = input("informe o produto a pesquisar o preço ou fim para sair: ").lower()

    if produto == "fim":
        break;
    if produto in produtos:
        print(f"produto {produto} custa {produtos[produto]}. ")
    else:
        print(f"produto {produto} não encontrado. ")
```
