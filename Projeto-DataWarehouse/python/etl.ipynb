```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
usuario = 'root'
senha = 'Jota/2025'
host = 'localhost'
banco = 'loja_koisinhas'

engine = create_engine(
    f'mysql+pymysql://{usuario}:{senha}@{host}/{banco}'
)

conn = engine.connect()

print("Conexão realizada com sucesso!")
```

```python
df_clientes = pd.read_csv('../csv/cliente.csv')
```

```python
df_clientes
```

```python
df_clientes = df_clientes.dropna()
df_clientes = df_clientes.drop_duplicates()
```

```python
df_clientes.to_sql(
    'dim_cliente',
    con=engine,
    if_exists='append',
    index=False
)
```
