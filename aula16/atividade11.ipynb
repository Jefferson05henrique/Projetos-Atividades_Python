```python
import pandas as pd
import numpy as np

# Criando um DataFrame inicial novamente para garantir que todos os passos sejam incluídos
data = {
    'A': [1, 2, np.nan, 4, 5, np.nan, 7, 8, 9, np.nan, 11, 12, np.nan, 14, 15, np.nan, 17, 18, np.nan, 20],
    'B': [np.nan, 1, 2, np.nan, 4, 5, 6, np.nan, 8, 9, 10, 11, np.nan, 13, 14, 15, np.nan, 17, 18, 19],
    'C': [1, 2, 3, 4, 5, np.nan, 7, np.nan, 9, 10, 11, 12, 13, 14, 15, np.nan, 17, 18, np.nan, 20]
}

df3 = pd.DataFrame(data)

# Duplicando o DataFrame
df3_duplicado = pd.concat([df3, df3])

# Contabilizando os valores nulos
valores_nulos = df3_duplicado.isnull().sum()
total_nulos = df3_duplicado.isnull().sum().sum()

print("Contabilização de valores nulos por coluna:")
print(valores_nulos)
print(f"\nTotal de valores nulos: {total_nulos}")

# Criando df4 (removendo duplicatas)
df4 = df3_duplicado.drop_duplicates()

# Criando df5 (removendo valores vazios)
df5 = df3_duplicado.dropna()

# Mostrando o shape e a descrição de df4
print("\nShape do df4 (sem duplicatas):", df4.shape)
print("\nDescrição do df4:")
print(df4.describe())

# Mostrando o shape e a descrição de df5
print("\nShape do df5 (sem valores vazios):", df5.shape)
print("\nDescrição do df5:")
print(df5.describe())
```

```python
import pandas as pd
import numpy as np

# Criando um DataFrame inicial
data = {
    'volume_hl': [1, 2, np.nan, 4, 5, np.nan, 7, 8, 9, np.nan, 11, 12, np.nan, 14, 15, np.nan, 17, 18, np.nan, 20],
    'value_usd': [np.nan, 1, 2, np.nan, 4, 5, 6, np.nan, 8, 9, 10, 11, np.nan, 13, 14, 15, np.nan, 17, 18, 19],
    'C': [1, 2, 3, 4, 5, np.nan, 7, np.nan, 9, 10, 11, 12, 13, 14, 15, np.nan, 17, 18, np.nan, 20]
}

df3 = pd.DataFrame(data)

print("Shape do df3 antes de remover duplicatas:", df3.shape)

# Adicionamos .copy() para garantir que dfFinal seja um objeto independente na memória
dfFinal = df3.drop_duplicates().copy()

print("Shape do dfFinal depois de remover duplicatas:", dfFinal.shape)

# Substituindo valores vazios usando ATRIBUIÇÃO (sem inplace=True)
dfFinal['volume_hl'] = dfFinal['volume_hl'].fillna(dfFinal['volume_hl'].median())
dfFinal['value_usd'] = dfFinal['value_usd'].fillna(dfFinal['value_usd'].median())

# Mostrando o dfFinal após substituição
print("\nDataFrame dfFinal após substituição dos valores vazios:")
print(dfFinal)
```

```python
import pandas as pd
import numpy as np

# 1. Preparação dos Dados Originais
data = {
    'volume_hl': [1, 2, np.nan, 4, 5, np.nan, 7, 8, 9, np.nan, 11, 12, np.nan, 14, 15, np.nan, 17, 18, np.nan, 20],
    'value_usd': [np.nan, 1, 2, np.nan, 4, 5, 6, np.nan, 8, 9, 10, 11, np.nan, 13, 14, 15, np.nan, 17, 18, 19],
    'C': [1, 2, 3, 4, 5, np.nan, 7, np.nan, 9, 10, 11, 12, 13, 14, 15, np.nan, 17, 18, np.nan, 20]
}
df3 = pd.DataFrame(data)

# 2. Processamento: Criando dfFinal com limpeza
dfFinal = df3.drop_duplicates().copy()
dfFinal['volume_hl'] = dfFinal['volume_hl'].fillna(dfFinal['volume_hl'].median())
dfFinal['value_usd'] = dfFinal['value_usd'].fillna(dfFinal['value_usd'].median())

# --- RELATÓRIO DE COMPARAÇÃO ---

print("="*30)
print("  COMPARAÇÃO DE DATAFRAMES")
print("="*30)

## --- DATAFRAME ORIGINAL (df3) ---
print("\n>>> DATAFRAME ORIGINAL (df3)")
print(f"Shape: {df3.shape}")
print("-" * 20)
print(df3.describe().T)

print("\n" + "="*50)

## --- DATAFRAME PROCESSADO (dfFinal) ---
print("\n>>> DATAFRAME FINAL (dfFinal)")
print(f"Shape: {dfFinal.shape}")
print("-" * 20)
print(dfFinal.describe().T)

print("\n" + "="*50)

# --- ANÁLISE DE MUDANÇAS ---
print("\n>>> RESUMO DAS ALTERAÇÕES:")
print(f"* Linhas removidas (Duplicatas): {df3.shape[0] - dfFinal.shape[0]}")
print(f"* Valores nulos em 'volume_hl': {df3['volume_hl'].isna().sum()} -> {dfFinal['volume_hl'].isna().sum()}")
print(f"* Valores nulos em 'value_usd': {df3['value_usd'].isna().sum()} -> {dfFinal['value_usd'].isna().sum()}")
```
