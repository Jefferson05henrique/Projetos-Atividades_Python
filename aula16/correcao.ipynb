```python
import pandas as pd

df = pd.read_excel('vinhos_tratamento.xlsx')
df2 = df.copy()
df3 = pd.concat([df2, df2])
print(df3.shape)
print("." * 50)
print(df3.describe())
```

```python
print(df3.isnull().sum())

df4 = df3.copy()
df4 = df4.drop_duplicates()

print(df4.shape)
print(df4.describe())
```

```python
df5 = df3.copy()
df5 = df5.drop_duplicates()
print("APRESENTAÇÃO DF5:")
print("-"*50)
print(df5.shape)
print("-"*50)
print(df5.describe())
print("-"*50)
```

```python
df_final = df3.copy()
print(df_final.shape)
df_final = df_final.drop_duplicates()
print(df_final.shape)
 
media_volume_hl = df_final["volume_hl"].mean()
media_value_usd = df_final["value_usd"].mean()
df_final["volume_hl"] = df_final["volume_hl"].fillna(media_volume_hl)
df_final["value_usd"] = df_final["value_usd"].fillna(media_value_usd)

print("-"*50)
print(df_final)
```
