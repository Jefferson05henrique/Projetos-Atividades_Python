```python
import pandas as pd

df = pd.read_excel('vinhos.xlsx')

medias = []
for i in range(5):
    amostra = df.sample(frac=0.05, random_state=i)
    media = amostra['volume_hl'].mean()
    medias.append(media)
    print(f"Amostra {i+1}: média de volume_hl = {round(media, 2)}")

print("Médias registradas: ", medias)
```

```python
import pandas as pd

df = pd.read_excel('vinhos.xlsx')

# Calcular a média da população para volume_hl
media_populacao_volume = df['volume_hl'].mean()
print(f"Média da população para volume_hl: {round(media_populacao_volume, 2)}")

# Comparar com as médias das amostras (assumindo que 'medias' foi calculado anteriormente)
if 'medias' in globals():
    print("\nComparação das médias das amostras com a média da população:")
    for i, media_amostra in enumerate(medias):
        diferenca = media_amostra - media_populacao_volume
        print(f"Amostra {i+1}: {round(media_amostra, 2)} (Diferença: {round(diferenca, 2)})")
else:
    print("Variável 'medias' não encontrada. Execute a célula anterior primeiro.")
```

Amostras podem variar devido à variabilidade inerente da população, ao processo aleatório de seleção e ao tamanho limitado da amostra, resultando em diferenças nas estatísticas calculadas.
