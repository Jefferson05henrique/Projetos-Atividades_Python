```python
import pandas as pd
import numpy as np

# Carregar o arquivo Excel
df = pd.read_excel('vinhos_exercicio.xlsx')

# Explorar a estrutura dos dados
print("Primeiras linhas do dataset:")
print(df.head())
print("\n" + "="*50)
print("Informações sobre o dataset:")
print(df.info())
print("\n" + "="*50)
print("Colunas disponíveis:")
print(df.columns.tolist())
```

```python
# Atividade 1: Frequência de Tipos de Uva
# Contar a frequência de cada tipo de uva
freq_absoluta = df['grape_type'].value_counts().sort_index()

# Calcular frequência relativa (%)
freq_relativa = (freq_absoluta / len(df) * 100).round(2)

# Calcular frequência acumulada
freq_acumulada = freq_absoluta.cumsum()

# Criar tabela com todas as informações
tabela_atividade1 = pd.DataFrame({
    'Frequência Absoluta': freq_absoluta,
    'Frequência Relativa (%)': freq_relativa,
    'Frequência Acumulada': freq_acumulada
})

print("ATIVIDADE 1: Frequência de Tipos de Uva (grape_type)")
print("="*60)
print(tabela_atividade1)
print("\n")
```

```python
# Atividade 2: Quartis de Valor em USD
# Dividir value_usd em 4 classes iguais (quartis) usando pd.qcut
quartis = pd.qcut(df['value_usd'], q=4, labels=['Q1 (25%)', 'Q2 (50%)', 'Q3 (75%)', 'Q4 (100%)'])

# Contar frequência absoluta de cada quartil
freq_absoluta_quartil = quartis.value_counts().sort_index()

# Calcular frequência relativa (%)
freq_relativa_quartil = (freq_absoluta_quartil / len(df) * 100).round(2)

# Criar tabela com as informações
tabela_atividade2 = pd.DataFrame({
    'Frequência Absoluta': freq_absoluta_quartil,
    'Frequência Relativa (%)': freq_relativa_quartil
})

print("ATIVIDADE 2: Quartis de Valor em USD (value_usd)")
print("="*60)
print(tabela_atividade2)
print("\nInterpretação:")
print("- Q1 (25%): Valores mais baixos de USD")
print("- Q2 (50%): Segundo quartil de valores")
print("- Q3 (75%): Terceiro quartil de valores")
print("- Q4 (100%): Valores mais altos de USD")
print("\n")
```

```python
# Atividade 3: Frequência de Anos por Faixas
# Criar uma função para categorizar os anos nas faixas especificadas
def categorizar_ano(ano):
    if 2014 <= ano <= 2018:
        return '2014-2018'
    elif 2019 <= ano <= 2022:
        return '2019-2022'
    elif 2023 <= ano <= 2027:
        return '2023-2027'
    else:
        return 'Fora da faixa'

# Aplicar a função para criar uma coluna de faixas
faixas_ano = df['year'].apply(categorizar_ano)

# Contar frequência absoluta de cada faixa
freq_absoluta_faixa = faixas_ano.value_counts().sort_index()

# Calcular frequência relativa (%)
freq_relativa_faixa = (freq_absoluta_faixa / len(df) * 100).round(2)

# Calcular frequência acumulada
freq_acumulada_faixa = freq_absoluta_faixa.cumsum()

# Criar tabela com as informações
tabela_atividade3 = pd.DataFrame({
    'Frequência Absoluta': freq_absoluta_faixa,
    'Frequência Relativa (%)': freq_relativa_faixa,
    'Frequência Acumulada': freq_acumulada_faixa
})

print("ATIVIDADE 3: Frequência de Anos (year) por Faixas")
print("="*60)
print(tabela_atividade3)
print("\n")
```
