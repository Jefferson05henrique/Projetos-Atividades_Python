```python
import pandas as pd

df = pd.read_csv('wine.csv')

print(df.head())

print("\n shape:", df.shape)

print("\n Dtypes:", df.dtypes)

print("\n Describe:", df.describe())
```

```python
print("\n Describle transporto:\n ", df.describe().T)
```

```python
import numpy as np

df_copy = df.copy()

for _ in range(20):
    i = np.random.randint(0, df_copy.shape[0])
    j = np.random.randint(0, df_copy.shape[1])
    df_copy.iat[i, j] = np.nan

print("\n Dados com valores faltantes:\n", df_copy.isnull().sum())
```

```python
print("\n Describe original:\n", df.describe().T)

print("\n Describe com nulos:\n", df_copy.describe().T)
```
