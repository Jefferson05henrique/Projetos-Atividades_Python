```python
import pandas as pd
import joblib
from sklearn.neighbors import KNeighborsClassifier
import io

print(pd.__version__)
print(joblib.__version__)
print(KNeighborsClassifier.__module__)
print(io.__name__)
```

```python
URL_DADOS = "https://docs.google.com/spreadsheets/d/e/2PACX-1vSiGX8co0hgwOm7obE-qVPoBxMDTJ2WaCNQcEuIuHqYeJ9i0uSR2W-29Mdb5G93nkP1-5DS1lfikJfw/pub?output=csv"

NOME_ARQUIVO_MODELO = 'Com lacunas/modelo_desempenho.pkl'
```

```python
data = pd.read_csv(URL_DADOS)

print(data.head())
```

```python
features = ['horas_estudo', 'faltas', 'nota_p1']
target = 'resultado'

X = data[features]
Y = data[target]
```

```python
print(X.info())
print(Y.info())
```

```python
modelo = KNeighborsClassifier(n_neighbors=3)

modelo.fit(X, Y)

print(modelo.classes_)
```

```python
joblib.dump(modelo, NOME_ARQUIVO_MODELO)

print("Modelo salvo!")
```

```python
def treinar_modelo():
    print(f"Baixando dados de {URL_DADOS}...")
    
    data = pd.read_csv(URL_DADOS)
    
    print("--- Dados Carregados ---")
    print(data.head())

    print("--- Preparando dados para o treino ---")
    features = ['horas_estudo', 'faltas', 'nota_p1']
    target = 'resultado'
    X = data[features]
    Y = data[target]
```

```python
modelo = KNeighborsClassifier(n_neighbors=3)
```

```python
print(X.info())
print(Y.info())

print(X.isnull().sum())
print(Y.isnull().sum())
```

```python
print(X.head())
print(Y.head())
print(X.shape)

modelo.fit(X, Y)

print(f"--- Modelo Treinado! Classes: {modelo.classes_} ---")
```

```python
joblib.dump(modelo, NOME_ARQUIVO_MODELO)
    
print(f"--- Modelo salvo com sucesso em '{NOME_ARQUIVO_MODELO}' ---")
```

```python
if __name__ == "__main__":
    treinar_modelo()
```
