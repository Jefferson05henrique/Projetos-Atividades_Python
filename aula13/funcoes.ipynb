```python
def tamanho_palavra(palavra):
    return len(palavra)  

resultado = tamanho_palavra("python")
print(f"A palavra tem {resultado} caracteres.")
```

```python
# Função para converter Fahrenheit para Celsius e Kelvin
def converter_temperatura(fahrenheit):
    celsius = (fahrenheit - 32) * 5/9
    kelvin = celsius + 273.15
    print(f"{fahrenheit}°F equivalem a {celsius:.2f}°C e {kelvin:.2f}K")

converter_temperatura(98.6)
```

```python
num = int(input("numero: "))

def par_impar(num):
    if num % 2 == 0:
        return "par"
    else:
        return "impar"
    
par_impar(num)
```
