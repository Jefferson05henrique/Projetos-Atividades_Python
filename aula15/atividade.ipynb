```python
import pandas as pd
# dataframe geral
df = pd.read_excel("vinhos_exercicio.xlsx")
# dataframe 1
df_wine = df[
    (df["grape_type"]=="Wine") &
    (df["location_type"]=="Brazil_State") &
    (df["category"]=="Sales")
    ]
# dataframe 2
df_table = df[
    (df["grape_type"]=="Table") &
    (df["category"]=="Sales")
    ]
desc_wine = df_wine.describe().T
desc_table = df_table.describe().T
media_wine = desc_wine.loc["value_usd","mean"]
media_table = desc_table.loc["value_usd","mean"]
print(f"Média Wine: {media_wine}")
print(f"Média Table: {media_table}")

if media_wine > media_table:
    print(f"Wine possui a maior média em value_usd!")
elif media_wine==media_table:
    print("Ambas tem a mesma média em value_usd!")
else:
    print("Table possui a maior média em value_usd!")
```

```python
df_wine_ord = df_wine.sort_values(by=["year","value_usd"], ascending=[False,True])
df_table_ord = df_table.sort_values(by=["year","value_usd"], ascending=[False,True])
df_wine_final = df_wine_ord.iloc[:,[0,1,4,5,6]]
df_table_final = df_table_ord.iloc[:,[0,1,4,5,6]]

print(df_wine_final)
print("="*50)
print(df_table_final)
print(df)
```

```python
df_wine_top10 = df_wine.reset_index(drop=True).loc[:9]
df_table_top10 = df_table.reset_index(drop=True).loc[:9]

print(df_wine_top10)
print("="*80)
print(df_table_top10)
```
