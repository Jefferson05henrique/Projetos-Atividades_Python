# 4️⃣ Visualizações Básicas com Matplotlib

## 📈 Gráfico 1: Distribuição de Vendas por Categoria

### 🎓 O que é um Gráfico de Barras?

Um **gráfico de barras** mostra:
- **Eixo X (horizontal)**: Categorias (Vendas, Importação, Exportação, etc)
- **Eixo Y (vertical)**: Valores (quanto em USD)
- **Altura das barras**: Quanto maior, maior o valor

**Como interpretar:**
- Compare a altura das barras
- Identifique qual tem maior/menor valor
- Veja as diferenças proporcionais

Vamos criar:

```python
# Calcular o total de vendas por categoria
vendas_por_categoria = df.groupby('categoria')['valor_usd'].sum().sort_values(ascending=False)

# Criar figura e eixos
fig, ax = plt.subplots(figsize=(10, 6))

# Criar o gráfico de barras
barras = ax.bar(vendas_por_categoria.index, vendas_por_categoria.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A'])

# Adicionar títulos e rótulos
ax.set_title('💰 Total de Vendas por Categoria de Vinho', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Categoria', fontsize=12, fontweight='bold')
ax.set_ylabel('Valor em USD', fontsize=12, fontweight='bold')

# Adicionar valores no topo das barras (para facilitar leitura)
for barra in barras:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2., altura,
            f'${altura:,.0f}',
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# Adicionar grade (linhas horizontais)
ax.grid(axis='y', alpha=0.3)

# Rotacionar rótulos do eixo X para melhor leitura
plt.xticks(rotation=45, ha='right')

# Ajustar layout para não cortar nada
plt.tight_layout()

# Exibir o gráfico
plt.show()

print("\n🔍 INSIGHTS DO GRÁFICO:")
print(f"  ✅ Categoria com MAIOR vendas: {vendas_por_categoria.idxmax()} (${vendas_por_categoria.max():,.2f})")
print(f"  ❌ Categoria com MENOR vendas: {vendas_por_categoria.idxmin()} (${vendas_por_categoria.min():,.2f})")
print(f"  📊 Total geral: ${vendas_por_categoria.sum():,.2f}")
```

## 📈 Gráfico 2: Evolução de Vendas ao Longo dos Anos

### 🎓 O que é um Gráfico de Linhas?

Um **gráfico de linhas** mostra:
- **Eixo X (horizontal)**: Tempo (anos)
- **Eixo Y (vertical)**: Valor em USD
- **Linhas conectando pontos**: Mostram a **tendência** ao longo do tempo

**Como interpretar:**
- Se a linha sobe → vendas aumentaram
- Se a linha desce → vendas diminuíram
- Se é reta → vendas estáveis
- A **inclinação** mostra a velocidade de mudança

```python
# Calcular vendas por ano
vendas_por_ano = df.groupby('ano')['valor_usd'].sum().sort_index()

# Criar figura
fig, ax = plt.subplots(figsize=(12, 6))

# Criar gráfico de linhas
ax.plot(vendas_por_ano.index, vendas_por_ano.values, marker='o', linewidth=3, 
        markersize=8, color='#2E86AB', markerfacecolor='#A23B72', markeredgewidth=2, markeredgecolor='#2E86AB')

# Adicionar títulos
ax.set_title('📈 Evolução de Vendas Ao Longo dos Anos', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Ano', fontsize=12, fontweight='bold')
ax.set_ylabel('Valor em USD', fontsize=12, fontweight='bold')

# Adicionar valores nos pontos
for ano, valor in vendas_por_ano.items():
    ax.text(ano, valor, f'${valor:,.0f}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# Adicionar grade
ax.grid(True, alpha=0.3)

# Formatar eixo X para mostrar apenas anos inteiros
ax.set_xticks(vendas_por_ano.index)

plt.tight_layout()
plt.show()

# Calcular crescimento
crescimento = vendas_por_ano.iloc[-1] - vendas_por_ano.iloc[0]
pct_crescimento = (crescimento / vendas_por_ano.iloc[0]) * 100

print("\n🔍 INSIGHTS DO GRÁFICO:")
print(f"  🏆 Melhor ano: {vendas_por_ano.idxmax()} (${vendas_por_ano.max():,.2f})")
print(f"  📉 Pior ano: {vendas_por_ano.idxmin()} (${vendas_por_ano.min():,.2f})")
print(f"  📊 Crescimento total: ${crescimento:,.2f} ({pct_crescimento:+.1f}%)")
```

## 🍇 Gráfico 3: Top 10 Tipos de Uva Mais Vendidos

### 🎓 O que é um Gráfico Horizontal?

É uma **variação do gráfico de barras**, onde:
- **Eixo Y (vertical)**: Categorias (tipos de uva)
- **Eixo X (horizontal)**: Valores (USD)

**Vantagem:** Melhor para muitas categorias com nomes longos!

```python
# Calcular top 10 tipos de uva
top_uvas = df.groupby('tipo_uva')['valor_usd'].sum().nlargest(10).sort_values()

# Criar figura com cores gradiente
fig, ax = plt.subplots(figsize=(10, 8))

# Criar gráfico de barras horizontal
cores = plt.cm.RdYlGn(np.linspace(0.3, 0.9, len(top_uvas)))
barras = ax.barh(range(len(top_uvas)), top_uvas.values, color=cores)

# Títulos e rótulos
ax.set_title('🍇 Top 10 Tipos de Uva Mais Vendidos', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Valor em USD', fontsize=12, fontweight='bold')
ax.set_ylabel('Tipo de Uva', fontsize=12, fontweight='bold')
ax.set_yticks(range(len(top_uvas)))
ax.set_yticklabels(top_uvas.index)

# Adicionar valores nas barras
for i, (barra, valor) in enumerate(zip(barras, top_uvas.values)):
    ax.text(valor, i, f' ${valor:,.0f}', va='center', fontweight='bold', fontsize=10)

# Adicionar grade
ax.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n🔍 INSIGHTS DO GRÁFICO:")
print(f"  🥇 Tipo de uva mais vendido: {top_uvas.idxmax()} (${top_uvas.max():,.2f})")
print(f"  🥈 Segunda mais vendida: {top_uvas.iloc[-2]} (${top_uvas.values[-2]:,.2f})")
print(f"  📊 Soma Top 10: ${top_uvas.sum():,.2f}")
```

# 5️⃣ Visualizações Avançadas com Seaborn

## 🔥 Gráfico 4: Mapa de Calor (Heatmap) - Correlações

### 🎓 O que é Correlação?

**Correlação** mede se **duas variáveis estão relacionadas**:
- **Correlação positiva** (verde): Quando uma aumenta, a outra aumenta
- **Correlação negativa** (vermelho): Quando uma aumenta, a outra diminui
- **Sem correlação** (branco): Não há relação

### 🎓 O que é um Heatmap?

Um **mapa de calor** mostra correlações através de **cores**:
- **Verde escuro** = Correlação forte positiva (1.0)
- **Branco** = Sem correlação (0.0)
- **Vermelho escuro** = Correlação forte negativa (-1.0)

**Por que usar?** Identifica rapidamente quais variáveis estão relacionadas!

```python
# Selecionar apenas colunas numéricas para correlação
colunas_numericas = df.select_dtypes(include=[np.number]).columns
matriz_correlacao = df[colunas_numericas].corr()

# Criar figura
fig, ax = plt.subplots(figsize=(10, 8))

# Criar heatmap
sns.heatmap(matriz_correlacao, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            square=True, linewidths=2, cbar_kws={"shrink": 0.8}, ax=ax, 
            annot_kws={'size': 11, 'weight': 'bold'})

# Títulos
ax.set_title('🔥 Mapa de Calor - Correlação Entre Variáveis Numéricas', 
             fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()

print("\n🔍 INSIGHTS DO GRÁFICO:")
print("\n📌 Correlações Importantes:")
# Encontrar correlações fortes (exceto a diagonal)
for i in range(len(matriz_correlacao.columns)):
    for j in range(i+1, len(matriz_correlacao.columns)):
        corr_valor = matriz_correlacao.iloc[i, j]
        if abs(corr_valor) > 0.5:  # Correlação moderada ou forte
            var1 = matriz_correlacao.columns[i]
            var2 = matriz_correlacao.columns[j]
            tipo = "Positiva ⬆️" if corr_valor > 0 else "Negativa ⬇️"
            print(f"  • {var1} ↔️ {var2}: {corr_valor:.3f} ({tipo})")
```

## 📊 Gráfico 5: Distribuição de Preços (Histograma)

### 🎓 O que é um Histograma?

Um **histograma** mostra:
- **Eixo X**: Faixa de preços (por exemplo: 0-100, 100-200, etc)
- **Eixo Y**: Quantos vinhos têm aquele preço
- **Altura das barras**: Frequência

**Como interpretar:**
- Se a forma é **em sino** → distribuição normal (equilibrada)
- Se tem **pico à esquerda** → mais produtos baratos
- Se tem **pico à direita** → mais produtos caros
- Se tem **2 picos** → existem 2 grupos distintos

```python
# Criar figura
fig, ax = plt.subplots(figsize=(12, 6))

# Criar histograma
n, bins, patches = ax.hist(df['preco_medio'], bins=30, color='#3498db', alpha=0.7, edgecolor='black')

# Adicionar linha de média
media = df['preco_medio'].mean()
mediana = df['preco_medio'].median()

ax.axvline(media, color='#e74c3c', linestyle='--', linewidth=2, label=f'Média: ${media:.2f}')
ax.axvline(mediana, color='#2ecc71', linestyle='--', linewidth=2, label=f'Mediana: ${mediana:.2f}')

# Títulos
ax.set_title('📊 Distribuição de Preços Médios dos Vinhos', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Preço Médio (USD por HL)', fontsize=12, fontweight='bold')
ax.set_ylabel('Quantidade de Vinhos', fontsize=12, fontweight='bold')

# Adicionar legenda
ax.legend(fontsize=11, loc='upper right')

# Adicionar grade
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n🔍 INSIGHTS DO GRÁFICO:")
print(f"  📊 Preço Médio: ${media:.2f}")
print(f"  📌 Preço Mediano: ${mediana:.2f}")
print(f"  📈 Preço Máximo: ${df['preco_medio'].max():.2f}")
print(f"  📉 Preço Mínimo: ${df['preco_medio'].min():.2f}")
print(f"  📐 Desvio Padrão: ${df['preco_medio'].std():.2f}")
print(f"  💡 Total de vinhos: {len(df)}")
```

## 📦 Gráfico 6: Boxplot - Comparar Distribuições por Categoria

### 🎓 O que é um Boxplot?

Um **boxplot** (caixa e bigodes) mostra:
- **Linha inferior (Q1)**: 25% dos dados abaixo
- **Linha no meio (Mediana)**: 50% dos dados abaixo
- **Linha superior (Q3)**: 75% dos dados abaixo
- **Bigodes**: Extensão dos dados normais
- **Pontos isolados**: Outliers (valores extremos)

**Para que serve:** Comparar distribuições rapidamente e identificar valores anormais!

```python
# Criar figura
fig, ax = plt.subplots(figsize=(12, 6))

# Criar boxplot
sns.boxplot(data=df, x='categoria', y='preco_medio', palette='Set2', ax=ax, width=0.6)

# Adicionar pontos individuais (mais leve)
sns.stripplot(data=df, x='categoria', y='preco_medio', color='black', alpha=0.3, size=4, ax=ax)

# Títulos
ax.set_title('📦 Distribuição de Preços por Categoria (Boxplot)', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Categoria', fontsize=12, fontweight='bold')
ax.set_ylabel('Preço Médio (USD por HL)', fontsize=12, fontweight='bold')

# Adicionar grade
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n🔍 INSIGHTS DO GRÁFICO:")
print("\n📊 Estatísticas por Categoria:\n")
estadisticas = df.groupby('categoria')['preco_medio'].agg(['count', 'mean', 'median', 'min', 'max', 'std'])
print(estadisticas.round(2))

print("\n💡 Interpretação:")
for categoria in df['categoria'].unique():
    dados_cat = df[df['categoria'] == categoria]['preco_medio']
    print(f"  • {categoria}: Preço médio ${dados_cat.mean():.2f} (variação de ${dados_cat.min():.2f} a ${dados_cat.max():.2f})")
```

## 🔵 Gráfico 8: Scatter Plot - Relação Volume vs Preço

### 🎓 O que é um Scatter Plot?

Um **scatter plot** (gráfico de dispersão) mostra:
- **Eixo X**: Uma variável (neste caso, volume)
- **Eixo Y**: Outra variável (neste caso, preço)
- **Pontos**: Cada vinho é um ponto
- **Padrão dos pontos**: Mostra se há relação entre as variáveis

**Como interpretar:**
- Se os pontos **sobem da esquerda para direita** → Correlação positiva
- Se os pontos **descem da esquerda para direita** → Correlação negativa
- Se os pontos **estão espalhados** → Pouca ou nenhuma correlação

```python
# Criar scatter plot interativo
fig = px.scatter(df, 
                 x='volume_hl', 
                 y='preco_medio', 
                 color='categoria',
                 size='valor_usd',
                 hover_data=['tipo_uva', 'ano'],
                 title='🔵 Relação Volume vs Preço Médio (colorido por Categoria)',
                 labels={'volume_hl': 'Volume (Hectolitros)', 'preco_medio': 'Preço Médio (USD/HL)'},
                 height=600)

# Personalizar
fig.update_layout(
    font=dict(size=11),
    title_font_size=16,
    hovermode='closest'
)

# Mostrar
fig.show()

print("\n🔍 INTERPRETAÇÃO:")
# Calcular correlação
corr = df['volume_hl'].corr(df['preco_medio'])
print(f"\n📊 Correlação entre Volume e Preço Médio: {corr:.3f}")

if abs(corr) < 0.3:
    interpretacao = "Muito fraca - sem relação clara"
elif abs(corr) < 0.7:
    interpretacao = "Moderada - existe alguma relação"
else:
    interpretacao = "Forte - existe relação clara"

print(f"   Interpretação: {interpretacao}")

if corr > 0:
    print(f"   ➡️ Relação: Vinhos com MAIS volume tendem a ter preço MAIOR")
else:
    print(f"   ➡️ Relação: Vinhos com MAIS volume tendem a ter preço MENOR")
```

# 7️⃣ Cálculo de KPIs (Indicadores Chave)

## 📊 O que é um KPI?

**KPI** = Key Performance Indicator (Indicador Chave de Desempenho)

São **números importantes** que mostram como o negócio está indo:
- 💰 **Receita Total**: Quanto ganhamos
- 📈 **Crescimento**: Estamos crescendo?
- 🏆 **Top Produtos**: Quais vendem mais
- 📉 **Média**: Qual é o valor "típico"

Vamos calcular os KPIs dos dados de vinho:

```python
# Calcular KPIs principais
print("=" * 80)
print("🎯 KPIs - INDICADORES PRINCIPAIS")
print("=" * 80)

# 1. Receita Total
receita_total = df['valor_usd'].sum()
print(f"\n💰 RECEITA TOTAL: ${receita_total:,.2f}")

# 2. Número de produtos
num_produtos = len(df)
print(f"📦 NÚMERO DE PRODUTOS: {num_produtos:,}")

# 3. Receita por produto
receita_por_produto = receita_total / num_produtos
print(f"📊 RECEITA MÉDIA POR PRODUTO: ${receita_por_produto:,.2f}")

# 4. Volume total
volume_total = df['volume_hl'].sum()
print(f"🧪 VOLUME TOTAL: {volume_total:,.0f} HL (Hectolitros)")

# 5. Preço médio
preco_medio = df['preco_medio'].mean()
preco_mediano = df['preco_medio'].median()
print(f"\n💵 ANÁLISE DE PREÇOS:")
print(f"   Preço Médio: ${preco_medio:.2f}/HL")
print(f"   Preço Mediano: ${preco_mediano:.2f}/HL")
print(f"   Preço Mínimo: ${df['preco_medio'].min():.2f}/HL")
print(f"   Preço Máximo: ${df['preco_medio'].max():.2f}/HL")

# 6. Análise por categoria
print(f"\n📊 DESEMPENHO POR CATEGORIA:")
receita_cat = df.groupby('categoria').agg({
    'valor_usd': 'sum',
    'preco_medio': 'mean'
}).sort_values('valor_usd', ascending=False)

for categoria, row in receita_cat.iterrows():
    pct = (row['valor_usd'] / receita_total) * 100
    print(f"   {categoria:15} - Receita: ${row['valor_usd']:>15,.2f} ({pct:>5.1f}%) | Preço Médio: ${row['preco_medio']:.2f}")

# 7. Crescimento ano a ano
print(f"\n📈 CRESCIMENTO ANO A ANO:")
vendas_ano = df.groupby('ano')['valor_usd'].sum().sort_index()
for i in range(1, len(vendas_ano)):
    ano_anterior = vendas_ano.index[i-1]
    ano_atual = vendas_ano.index[i]
    valor_anterior = vendas_ano.iloc[i-1]
    valor_atual = vendas_ano.iloc[i]
    crescimento = valor_atual - valor_anterior
    pct_crescimento = (crescimento / valor_anterior) * 100
    
    seta = "📈" if crescimento > 0 else "📉"
    print(f"   {ano_anterior} → {ano_atual}: {seta} {pct_crescimento:+.1f}% (${crescimento:+,.2f})")

# 8. Top 3 cidades
print(f"\n🏆 TOP 3 LOCAIS DE PRODUÇÃO:")
top_locais = df.groupby('local')['valor_usd'].sum().nlargest(3)
for i, (local, valor) in enumerate(top_locais.items(), 1):
    pct = (valor / receita_total) * 100
    print(f"   {i}. {local}: ${valor:,.2f} ({pct:.1f}%)")
```
