```python
import pandas as pd
import numpy as np

df = pd.read_excel('vinhos_desafio.xlsx')    

df

# Análise da tabela inteira
```

```python
df.head(10) #aqui mostra as primeiras linhas, neste caso pedir ás 10 primeiras
```

```python
df.info() #aqui mostra o tipo de cada coluna, a quantidade de valores não nulos e a quantidade total de linhas
df.describe()  #aqui mostra a média, o desvio padrão, o valor mínimo, os quartis e o valor máximo de cada coluna numérica
df.columns     #aqui mostra o nome de cada coluna
```

```python
df.isnull().sum()       #aqui mostra a quantidade de valores nulos em cada coluna
df = df.dropna()    #aqui remove as linhas que possuem valores nulos
df = df.drop_duplicates()   #aqui remove as linhas duplicadas, ou seja, as linhas que possuem os mesmos valores em todas as colunas

df['valor_usd'] = pd.to_numeric(df['valor_usd'], errors='coerce')
df['volume_hl'] = pd.to_numeric(df['volume_hl'], errors='coerce')

#  pd.to_numeric(df['valor_usd'], errors='coerce')
#  Existe números em tabelas que vem em forma de texto ou sinais, e o pd.to_numeric() converte esses valores para números, e o errors='coerce' faz com que os valores que não puderem ser convertidos sejam transformados em NaN (Not a Number), ou seja, valores nulos.
```

```python
df ['preco_medio'] = df['valor_usd'] / df['volume_hl'] #aqui cria uma nova coluna chamada 'preco_medio' que é o resultado da divisão da coluna 'valor_usd' pela coluna 'volume_hl'

print(df.head()) #aqui mostra as primeiras linhas da tabela, agora com a nova coluna 'preco_medio' criada
print(df['preco_medio'].describe()) #aqui mostra a média, o desvio padrão, o valor mínimo, os quartis e o valor máximo da coluna 'preco_medio'
```

```python
pd.options.display.float_format = '${:,.2f}'.format #aqui formata os números para serem exibidos com o símbolo de dólar e com duas casas decimais

df.groupby('categoria')['valor_usd'].sum() #aqui agrupa a tabela pela coluna 'categoria' e soma os valores da coluna 'valor_usd' para cada categoria, ou seja, mostra o total de valor em dólares para cada categoria de vinho
```

```python
pd.options.display.float_format = '${:,.2f}'.format #aqui formata os números para serem exibidos com o símbolo de dólar e com duas casas decimais

df[df['categoria'] != 'vendas'].groupby('categoria')['valor_usd'].sum() #aqui agrupa a tabela pela coluna 'categoria' e soma os valores da coluna 'valor_usd' para cada categoria, mas somente para as categorias que não são 'vendas', ou seja, mostra o total de valor em dólares para cada categoria de vinho, exceto para a categoria 'vendas'
```

```python
pd.options.display.float_format = '${:,.2f}'.format

df.groupby('ano')['valor_usd'].sum() #aqui agrupa a tabela pela coluna 'ano' e soma os valores da coluna 'valor_usd' para cada ano, ou seja, mostra o total de valor em dólares para cada ano de produção dos vinhos
```

```python
pd.options.display.float_format = '${:,.2f}'.format

df.groupby('local')['valor_usd'].sum().sort_values(ascending=False) #aqui agrupa a tabela pela coluna 'local' e soma os valores da coluna 'valor_usd' para cada local, ou seja, mostra o total de valor em dólares para cada local de produção dos vinhos, e depois ordena os resultados em ordem decrescente, ou seja, do local com maior valor em dólares para o local com menor valor em dólares
```

```python
df.groupby('tipo_uva')['valor_usd'].sum() #aqui agrupa a tabela pela coluna 'tipo_uva' e soma os valores da coluna 'valor_usd' para cada tipo de vinho, ou seja, mostra o total de valor em dólares para cada tipo de vinho
```

```python
totais_ano = df.groupby('ano')['valor_usd'].sum().reset_index(name='total_valor_usd')
totais_categoria = df.groupby('categoria')['valor_usd'].sum().reset_index(name='total_valor_usd')
with pd.ExcelWriter('vinhos_tratado.xlsx') as writer:
    df.to_excel(writer, sheet_name='dados', index=False)
    totais_ano.to_excel(writer, sheet_name='totais_ano', index=False)
    totais_categoria.to_excel(writer, sheet_name='totais_categoria', index=False)
```

```python
df.to_excel('vinhos_tratado.xlsx', index=False) #aqui salva a tabela tratada em um novo arquivo Excel chamado 'vinhos_tratado.xlsx', e o index=False faz com que a coluna de índice não seja salva no arquivo, ou seja, somente as colunas da tabela serão salvas.
```
