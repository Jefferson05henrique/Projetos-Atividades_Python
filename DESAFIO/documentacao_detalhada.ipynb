### 2️⃣ `df = df.drop_duplicates()`
**O que faz:**
- **`drop_duplicates()`**: Remove linhas **exatamente iguais**
- **`df =`**: Salva o resultado de volta em df

**Exemplo:**
Se você tem:
```
   categoria   local    tipo_uva    ano
0  vendas     França   Cabernet    2020
1  vendas     França   Cabernet    2020  ← Linha duplicada!
2  vendas     Itália   Merlot      2020
```

**Por que fazer?**

- Pode ter sido erro de importação

**Importante:**

### 3️⃣ `df['valor_usd'] = pd.to_numeric(df['valor_usd'], errors='coerce')`
**O que faz:**
- **`pd.to_numeric()`**: Converte uma coluna para números
- **`errors='coerce'`**: Se não conseguir converter, transforma em NaN (vazio)
- **`df['valor_usd'] =`**: Salva o resultado na coluna

**Por que fazer?**
- Às vezes, números vêm como **texto** do Excel
- Exemplo: `'5000'` (texto) em vez de `5000` (número)
- Precisamos converter para fazer cálculos

**Exemplo de conversão:**
```
Antes:  '5000', '3000.50', 'invalido', '4500'
Depois: 5000.0, 3000.50, NaN (porque 'invalido' não é número), 4500.0
```

**O parâmetro `errors='coerce'`:**
- **`'coerce'`**: Converte erros em NaN (não quebra o código)
- **`'raise'`**: Dá erro se algo não é número (padrão)
- **`'ignore'`**: Deixa como estava se não conseguir converter

### 4️⃣ `df['volume_hl'] = pd.to_numeric(df['volume_hl'], errors='coerce')`
**Mesma explicação da linha anterior, mas para a coluna `volume_hl`**

- **`dropna()`**: Remove linhas com valores vazios (NaN)
- **`subset=['valor_usd', 'volume_hl']`**: Mas SÓ verifica essas 2 colunas

```
Antes:
   categoria   valor_usd    volume_hl
0  vendas      5000.0       100.0      ✅ Mantém
1  vendas      NaN          80.0       ❌ Remove (valor_usd vazio)
2  vendas      4500.0       NaN        ❌ Remove (volume_hl vazio)
3  vendas      3000.0       120.0      ✅ Mantém

Depois: Só linhas 0 e 3 continuam
```

**Por que nessas colunas específicas?**

## 📖 Explicação Detalhada do GROUPBY:

### ⚙️ O Conceito: GROUPBY
**O que é groupby?**
- Agrupa linhas por uma **coluna em comum**
- Depois realiza uma **operação** em cada grupo

**Analogia:**
Imagine que você tem contas bancárias:
```
Nome      Banco     Saldo
João      Banco A   1000
Maria     Banco A   2000
João      Banco B   500
Maria     Banco B   1500
```

`groupby('Banco')` cria 2 grupos:
- **Banco A**: João e Maria
- **Banco B**: João e Maria

Depois você pode **somar** o saldo de cada banco:
- **Banco A**: 1000 + 2000 = 3000
- **Banco B**: 500 + 1500 = 2000

---

### 1️⃣ `pd.options.display.float_format = '${:,.2f}'.format`
**O que faz:**
- **Formata** como os números aparecem na tela
- `$` = Adiciona símbolo de dólar
- `:,.2f` = Com vírgula (separador de milhares) e 2 casas decimais

**Exemplo:**
```
Antes:  1234567.89012345
Depois: $1,234,567.89
```

**Componentes de `'${:,.2f}'`:**
- `$` = Símbolo do dólar no início
- `:` = Marca o início da formatação
- `,` = Adiciona separador de milhares (1,000 em vez de 1000)
- `.2f` = 2 casas decimais, tipo float (número decimal)

**Importante:** É apenas visual! Não muda o número real, só como aparece.

---

### 2️⃣ `df.groupby('categoria')['valor_usd'].sum()`
**O que faz - passo a passo:**

**Passo 1: `groupby('categoria')`**
- Agrupa todas as linhas que têm a **mesma categoria**
- Exemplo: agrupa todas as "vendas", todas as "importações", etc.

**Passo 2: `['valor_usd']`**
- Seleciona a coluna que vamos operar
- Vamos trabalhar com a coluna `valor_usd`

**Passo 3: `.sum()`**
- **Soma** todos os valores de cada grupo
- Para cada categoria, soma todos os `valor_usd`

**Exemplo visual:**
```
Dados originais:
   categoria    valor_usd
0  vendas       5000
1  vendas       3000
2  importação   4500
3  vendas       4000
4  importação   2000

Depois de groupby('categoria')['valor_usd'].sum():
categoria
vendas         12000   (5000 + 3000 + 4000)
importação     6500    (4500 + 2000)
```

**Resultado:**
```
categoria
vendas          $12,000.00
importação      $6,500.00
```

---

### `df.to_excel()`
**O que faz:**
- **`.to_excel()`**: Exporta (salva) o dataframe em um arquivo Excel
- **`'vinhos_tratado.xlsx'`**: Nome do arquivo a criar/sobrescrever
- **`index=False`**: Não salva a coluna de índice (números 0, 1, 2, etc.)

**O parâmetro `index`:**
- **`index=False`**: Não inclui a coluna de números sequenciais
  ```
  Resultado no Excel (SEM a coluna de índice):
  categoria    local    tipo_uva    ano    valor_usd    volume_hl    preco_medio
  vendas       França   Cabernet    2020   5000         100          50.00
  ```

- **`index=True`** (padrão): Inclui a coluna de índice
  ```
  Resultado no Excel (COM a coluna de índice):
      categoria    local    tipo_uva    ano    valor_usd    volume_hl    preco_medio
  0   vendas       França   Cabernet    2020   5000         100          50.00
  1   vendas       Itália   Merlot      2020   3000         80           37.50
  ```

**Geralmente usamos `index=False`** porque esses números não são dados reais.

## 📖 Exportação Avançada (Múltiplas Abas):

```python
totais_ano = df.groupby('ano')['valor_usd'].sum().reset_index(name='total_valor_usd')
totais_categoria = df.groupby('categoria')['valor_usd'].sum().reset_index(name='total_valor_usd')

with pd.ExcelWriter('vinhos_tratado.xlsx') as writer:
    df.to_excel(writer, sheet_name='dados', index=False)
    totais_ano.to_excel(writer, sheet_name='totais_ano', index=False)
    totais_categoria.to_excel(writer, sheet_name='totais_categoria', index=False)
```

### Entendendo Cada Parte:

#### 1️⃣ `.reset_index(name='total_valor_usd')`
**O que faz:**
- Quando você faz `groupby().sum()`, o resultado é um índice nomeado
- **`.reset_index()`**: Transforma esse índice em uma coluna normal
- **`name='total_valor_usd'`**: Nomeia a coluna de valores

**Exemplo:**
```
Antes (é um índice):
ano
2020    50000
2021    60000

Depois (é uma coluna):
   ano    total_valor_usd
0  2020   50000
1  2021   60000
```

**Por que fazer?**
- Quando exportar para Excel, fica melhor organizado
- Como uma tabela normal, não como um índice

#### 2️⃣ `with pd.ExcelWriter('vinhos_tratado.xlsx') as writer:`
**O que faz:**
- **`ExcelWriter`**: Cria um "escritor" de arquivos Excel
- **`with ... as writer:`**: Abre o arquivo para escrever múltiplas abas
- **`with`**: Garante que o arquivo é fechado corretamente ao final

**Analogia:** É como abrir um arquivo para edição, adicionar várias coisas, e depois fechar.

#### 3️⃣ Adicionando Múltiplas Abas:
```python
df.to_excel(writer, sheet_name='dados', index=False)
totais_ano.to_excel(writer, sheet_name='totais_ano', index=False)
totais_categoria.to_excel(writer, sheet_name='totais_categoria', index=False)
```

**`sheet_name='...'`**: Nome da aba no Excel
- Primeira linha → aba chamada "dados" (dados completos)
- Segunda linha → aba chamada "totais_ano" (resumo por ano)
- Terceira linha → aba chamada "totais_categoria" (resumo por categoria)

**Resultado no Excel:**
```
[dados] [totais_ano] [totais_categoria]  ← Abas no topo
```

**Vantagem:** Tudo em um arquivo, mas dados organizados em abas diferentes!
